```python
#1. Prompt chaining for a customer support AI
#Goal: Build a simple prompt chain to simulate a customer service flow


def classify_support_message(message):
    """
    Classifies customer support messages into one of three categories:
    billing, technical, or general

    Args:
        message (str): The customer support message to classify

    Returns:
        str: Classification result in format 'category: <category>'
    """
    message_lower = message.lower()

    # Billing keywords
    billing_keywords = [
        'charge', 'invoice', 'payment', 'bill', 'refund',
        'price', 'cost', 'subscription', 'credit card', 'paid'
    ]

    # Technical keywords
    technical_keywords = [
        'error', 'bug', 'crash', 'not working', 'broken', 'issue',
        'login', 'password', 'access', 'load', 'slow', 'feature'
    ]

    # Check for billing keywords
    billing_score = sum(1 for keyword in billing_keywords if keyword in message_lower)

    # Check for technical keywords
    technical_score = sum(1 for keyword in technical_keywords if keyword in message_lower)

    # Classify based on scores
    if billing_score > technical_score:
        return "category: billing"
    elif technical_score > billing_score:
        return "category: technical"
    else:
        return "category: general"


# Classify the given message
message = "I was charged twice on my last invoice."
result = classify_support_message(message)
print(result)
```
